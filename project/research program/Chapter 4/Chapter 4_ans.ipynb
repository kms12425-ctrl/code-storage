> *   **Backpropagation:** The loss is the starting point. We calculate its gradients relative to each weight, telling the optimizer how to adjust weights to reduce future errors.

### Question: Parameters vs. Hyperparameters
**Definition of a Hyperparameter:**
- A hyperparameter is an external configuration variable whose value is set *by the user before the training process begins*. Its optimal value cannot be directly mathematically learned from the data itself.

**Difference between Parameters and Hyperparameters:**
- **Parameters:** Internal variables that the model learns and updates *automatically* from the data during training (e.g., neuron weights and biases).
- **Hyperparameters**: External settings determined *manually* by the researcher to control how the model learns (e.g., learning rate, number of hidden layers, batch size, epochs).

*(Reference Images)*  
<img src="img/Hyperparameters1.png" width="400"> <img src="img/Hyperparameters2.png" width="400">

```python
# Programming Exercise: Experiment with different hyperparameters

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt


# 1. Load CIFAR-10 dataset
# Use the dataset prepared in Chapter 2 (including train, validation, and test splits)
(train_images_full, train_labels_full), (test_images,
                                         test_labels) = tf.keras.datasets.cifar10.load_data()

# Normalize pixel values to be between 0 and 1
train_images_full = train_images_full / 255.0
test_images = test_images / 255.0

# Split into train and validation sets (e.g., 80% train, 20% validation)
train_images, val_images, train_labels, val_labels = train_test_split(
    train_images_full, train_labels_full, test_size=0.2, random_state=42
)


def create_and_train_model(learning_rate, batch_size, epochs):
    print(
        f"\n--- Training with LR: {learning_rate}, Batch Size: {batch_size}, Epochs: {epochs} ---")

    # 1. Redefine the model (ensure each experiment starts from scratch, not continuing from previous weights)
    model = Sequential([
        Flatten(input_shape=(32, 32, 3)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])

    # 2. Compile the model: pass the specific learning_rate here
    model.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    # 3. Train the model: pass specific batch_size and epochs here
    history = model.fit(
        train_images,
        train_labels,
        validation_data=(val_images, val_labels),
        epochs=epochs,
        batch_size=batch_size,
        verbose=0  # Set to 0 to disable progress bar printing, only view final results
    )

    # Print final epoch results
    final_train_acc = history.history['accuracy'][-1]
    final_val_acc = history.history['val_accuracy'][-1]
    print(
        f"Final Train Accuracy: {final_train_acc:.4f} | Final Val Accuracy: {final_val_acc:.4f}")

    return history


# --- Experiment 1: Different Learning Rates (fixed batch_size=64, epochs=15) ---
history_lr_large = create_and_train_model(
    learning_rate=0.1, batch_size=64, epochs=15)
history_lr_good = create_and_train_model(
    learning_rate=0.001, batch_size=64, epochs=15)
history_lr_small = create_and_train_model(
    learning_rate=0.00001, batch_size=64, epochs=15)

# --- Experiment 2: Different Batch Sizes (fixed lr=0.001, epochs=15) ---
history_batch_small = create_and_train_model(
    learning_rate=0.001, batch_size=32, epochs=15)
history_batch_large = create_and_train_model(
    learning_rate=0.001, batch_size=256, epochs=15)


# Visualization: Compare the 5 experiments
plt.figure(figsize=(14, 6))

# 1. Validation Accuracy Comparison
plt.subplot(1, 2, 1)
plt.plot(history_lr_large.history['val_accuracy'], label='LR=0.1 (Batch 64)')
plt.plot(history_lr_good.history['val_accuracy'], label='LR=0.001 (Batch 64)')
plt.plot(history_lr_small.history['val_accuracy'],


         label='LR=0.00001 (Batch 64)')
plt.plot(

    history_batch_small.history['val_accuracy'], label='Batch=32 (LR 0.001)')
plt.plot(
    history_batch_large.history['val_accuracy'], label='Batch=256 (LR 0.001)')
plt.title('Validation Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# 2. Validation Loss Comparison
plt.subplot(1, 2, 2)
plt.plot(history_lr_large.history['val_loss'], label='LR=0.1 (Batch 64)')
plt.plot(history_lr_good.history['val_loss'], label='LR=0.001 (Batch 64)')
plt.plot(history_lr_small.history['val_loss'], label='LR=0.00001 (Batch 64)')
plt.plot(history_batch_small.history['val_loss'], label='Batch=32 (LR 0.001)')
plt.plot(history_batch_large.history['val_loss'], label='Batch=256 (LR 0.001)')
plt.title('Validation Loss Comparison')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 4.1 Regularization Techniques

To improve generalization and prevent overfitting, you can use techniques like dropout and weight regularization.

**Tasks:**
1. Research **dropout layers**. What do they do during training? How can they help prevent overfitting?
2. Research **L2 regularization** (also called weight decay). What effect does it have on the model’s weights?

**Programming Exercise:**

* Add L2 regularization to one of the dense layers (e.g., Dense(units, kernel_regularizer=tf.keras.regularizers.l2(0.01))).
* Retrain the model and compare the training and validation accuracy before and after adding these regularization techniques.

> **Answer:**
> 
> **1. Dropout layers:** During training, dropout randomly deactivates (sets to zero) a fraction of neurons in a layer at each update step. This prevents overfitting by keeping neurons from co-adapting too much, forcing the network to learn more robust, redundant features instead of relying on a few specific nodes.
> 
> **2. L2 regularization:** L2 regularization (weight decay) adds a penalty term proportional to the square of the layer's weights to the loss function. This pushes the model's weights to be smaller and closer to zero, resulting in a simpler, smoother model that is less sensitive to noise in the training set.


```python
# Programming Exercise: Add regularization techniques

from tensorflow.keras.layers import Dropout, Flatten, Dense
from tensorflow.keras.regularizers import l2
from tensorflow.keras.models import Sequential
from tensorflow.keras.optimizers import Adam

print("\n--- Training Model with Dropout and L2 Regularization ---")

# Define the model with L2 regularization and Dropout
model_reg = Sequential([
    Flatten(input_shape=(32, 32, 3)),
    # Apply L2 Regularization
    Dense(128, activation='relu', kernel_regularizer=l2(0.01)),
    # Apply Dropout layer (50% probability)
    Dropout(0.5),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# Compile the model
model_reg.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history_reg = model_reg.fit(
    train_images,
    train_labels,
    validation_data=(val_images, val_labels),
    epochs=15,
    batch_size=64,
    verbose=1  # Set verbose=1 to observe the training progress
)

# Output the final results
final_train_acc_reg = history_reg.history['accuracy'][-1]
final_val_acc_reg = history_reg.history['val_accuracy'][-1]
print(
    f"Regularized Model -> Final Train Accuracy: {final_train_acc_reg:.4f} | Final Val Accuracy: {final_val_acc_reg:.4f}")
```

## 4.2 Regularization Techniques (Dropout vs. L2 Regularization):
- [You can insert related screenshots here if you have them]
- **Dropout Layer:** Randomly turns off (sets to zero) a fraction of neurons during training. **Pros**: Forces the network to learn robust features and strongly prevents overfitting. **Cons**: Increases training time.
- **L2 Regularization:** Adds a weight penalty (squared magnitude of weights) to the loss function. **Pros**: Keeps all weights small, restricting model complexity. **Cons**: Requires tuning the penalty parameter ($\lambda$).
- **Key Differences:** Dropout prevents overfitting by dynamically turning off neurons (changing network structure). L2 prevents overfitting by mathematically forcing weights to stay small (changing the loss equation).

### Question 1: Mathematical Formulas of Loss Functions
**Mean Squared Error (MSE):**
- Formula: $MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$
- <img src="img/MSE.png" width="600">
- Explanation: It measures the average squared difference between the predicted values ($\hat{y}$) and the true values ($y$). It is best suited for **continuous regression tasks** (e.g., predicting prices) because it penalizes large errors.

**Cross-Entropy Loss:**
- Formula: $CE = - \sum_{i=1}^{C} y_i \log(\hat{y}_i)$
- Explanation: It measures the difference between two probability distributions. It heavily penalizes models that are "confidently wrong" by using the $-\log$ function. It is best suited for **classification tasks**.

- **Explanation**: Gradient descent is an optimization algorithm used to minimize the loss function. It calculates the gradient (slope) of the loss and takes small steps in the opposite direction to find the lowest possible error.

**2.2 Binary Cross-Entropy Loss:**  
<img src="img/Cross-Entropy%20Loss.png" width="400">

- **Explanation:** The math equation relies on $-\log(p)$ if the true label is 1, and $-\log(1-p)$ if the true label is 0. If the true label is 1 but the model predicts a probability near 0, $-\log(0)$ goes to infinity, resulting in a massive penalty.

**2.3 Backpropagation:**  
<img src="img/Backpropagation1.png" width="400"> <img src="img/Backpropagation2.png" width="400">
- **Explanation:** Backpropagation calculates the gradient of the loss function with respect to each weight. It uses the "chain rule" from calculus to pass the error *backwards* from the output layer to the input layer, telling the model how to adjust weights to reduce the loss.
